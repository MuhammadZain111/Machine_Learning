# Calculations
Add a Total and Average column for each student.

```python
students.sum()
```

```python
students.mean(numeric_only=True)
```

Adding the Average Column to the Students Dataset

```python
students['Average'] =students[['Math', 'Science', 'Computer','English', 'Urdu']].mean(numeric_only=True,axis=1)
```

```python
students.describe()
```

```python
students['Total'] =students[['Math', 'Science', 'Computer' ,'English', 'Urdu', ]].sum(axis=1)
```

```python
students.describe()
```

```python
students['Grade'] = np.select(
    [
        students['Average'] >= 80,
        students['Average'] >= 70,
        students['Average'] >= 60,
        students['Average'] >= 50
    ],
    ['A', 'B', 'C', 'D'],
    default='F'
)
```

```python
students.describe()
```

```python
print(students.columns)
```

```python
print(students[['Average', 'Grade']].head())
```

```python
students = students.drop('average', axis=1)
```

```python
students.describe(include='all')
```

```python
students[students['Name'] == 'Ali Malik']
```

```python
students
```

```python
students['Pass/Fail'] = np.where(
    students['Average'] >= 50,
    'Pass',
    'Fail'
)
```

```python
students
```

```python
students['Rank'] = students['Total'].rank(
    ascending=False,
    method='min'
).astype(int)

students = students.sort_values('Rank')

students[['Name', 'Total', 'Rank']]
```

## **Identify "at-risk" students: attendance below 65% or average below 50%.**

```python
at_risk = students[
    (students['Attendance_Pct'] < 65) |
    (students['Average'] < 50)
]

at_risk
```

# Part 4: Visualization (at least 5 charts)

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
subject_avg = students[['Math', 'Science', 'English']].mean()

plt.figure(figsize=(8, 5))
subject_avg.plot(kind='bar')

plt.title('Average Marks per Subject')
plt.xlabel('Subject')
plt.ylabel('Average Marks')
plt.xticks(rotation=0)
plt.show()
```

```python
students
```

```python
grade_counts = students['Grade'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    grade_counts,
    labels=grade_counts.index,
    autopct='%1.1f%%'
)

plt.title('Grade Distribution')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Section',
    y='Math',
    data=students
)

plt.title('Math Marks by Section')
plt.xlabel('Section')
plt.ylabel('Math Marks')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x='Study_Hours_PerDay',
    y='Average',
    data=students
)

plt.title('Study Hours vs Average Marks')
plt.xlabel('Study Hours')
plt.ylabel('Average Marks')

plt.show()
```

```python
correlation = students[
    ['Attendance_Pct', 'Study_Hours_PerDay', 'Assignment_Score', 'Average']
].corr()
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix')
plt.show()
```
