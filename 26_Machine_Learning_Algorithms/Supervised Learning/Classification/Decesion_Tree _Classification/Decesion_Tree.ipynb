```python
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
```

```python
df.head()
```

```python
#independent feature and dependent features
X=df.iloc[:,:-1]
y=iris.target
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
     X, y, test_size=0.33, random_state=42)
```

```python
X_train
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
#prunning
treemodel = DecisionTreeClassifier(max_depth=3)
```

```python
treemodel.fit(X_train,y_train)
```

```python
from sklearn import tree
import matplotlib.pyplot as plt

plt.figure(figsize=(10,10))
tree.plot_tree(treemodel, filled=True)
```

```python
#prediction
y_pred=treemodel.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,accuracy_score,confusion_matrix,classification_report

score = accuracy_score(y_pred,y_test)
print(score)


```

```python
print(classification_report(y_pred,y_test))

```

```python
print (confusion_matrix(y_pred,y_test))
```

```python
print(treemodel.score(X_train,y_train))
print(treemodel.score(X_test,y_test))
```
