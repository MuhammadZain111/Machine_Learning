```python
import numpy as np
```

```python
X = np.array([
    [2.5, 40],
    [3.0, 50],
    [3.5, 70],
    [3.8, 85],
    [3.9, 90]
])
```

```python
y = np.array([0, 0, 1, 1, 1])
```

```python

# here we will perform the featire Scaling


# Calculate mean of each feature

mean = np.mean(X, axis=0)

# Calculate standard deviation

std = np.std(X, axis=0)

# Standardize training data

# Lets apply the Z Score Standardization


X_scaled = (X - mean) / std


# 3. EUCLIDEAN DISTANCE

def euclidean_distance(point1, point2):

    distance = np.sqrt(
        np.sum((point1 - point2) ** 2)
    )

    return distance


# 4. KNN ALGORITHM


def knn(X, y, new_point, k):

    # Store distances

    distances = []

    # Calculate distance from new point
    # to every training data point

    for i in range(len(X)):

        distance = euclidean_distance(
            X[i],
            new_point
        )

        distances.append(distance)

    # Convert distances to NumPy array

    distances = np.array(distances)

    # Find indices of nearest K neighbors

    k_indices = np.argsort(distances)[:k]

    # Get labels of nearest neighbors

    k_labels = y[k_indices]

    # Count votes for each class

    values, counts = np.unique(
        k_labels,
        return_counts=True
    )

    # Select class with maximum votes

    prediction = values[np.argmax(counts)]

    return prediction


# 5. NEW DATA POINT

# New student:
# CGPA = 3.4
# Marks = 65

new_student = np.array([3.4, 65])


# 6. SCALE NEW DATA


# Use training mean and standard deviation

new_student_scaled = (
    new_student - mean
) / std



# 7. MAKE PREDICTION


k = 3

prediction = knn(
    X_scaled,
    y,
    new_student_scaled,
    k
)


# 8. DISPLAY RESULT


print("New Student:", new_student)

print("K:", k)

print("Prediction:", prediction)

if prediction == 1:

    print("Student will get placed")

else:

    print("Student will not get placed")
```
